### Input:

```python
# import and preview input
input = pd.read_csv('input.csv', index_col = False)
df = pd.DataFrame(input)
df
```

```python
df["id"] = df.index + 1
df
```

```python
df_stack = df.set_index(["id", "RM Category", "Year", "Month"])
df_stack
```

```python
df_stack.isna().sum()
```

```python
df_stack = df_stack.fillna(0)
df_stack.isna().sum()
```

```python
df_stack.tail(10)
```

```python
df_stack = df_stack.stack()
```

```python
df_stack
```

```python
df
```

```python
df_melt = pd.melt(df, id_vars=["id", "RM Category", "Year", "Month"], var_name="Name", value_name="Value")
```

```python
df_melt = df_melt.fillna(0)
df_melt.isna().sum()
```

```python
df_melt
```

```python
df_melt.sort_values(by=["Name", "id"], inplace=True)
df_melt
```

```python
df_melt["3m_avg"] = df_melt["Value"].rolling(window=3).mean()
df_melt
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)
df_melt
```

```python
df_melt["6m_avg"] = df_melt["Value"].rolling(window=6).mean()
df_melt
```
