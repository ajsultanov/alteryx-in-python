### Input:

```python
# import and preview input
input = pd.read_csv('input.csv', index_col = False)
df = pd.DataFrame(input)
df
```

```python
# assign record ids

df["id"] = df.index + 1
df
```

```python
# melt unpivots a dataframe from wide to long format

df = pd.melt(df, id_vars=["id", "RM Category", "Year", "Month"], var_name="Name", value_name="Value")
df
```

```python
# fillna replaces NaN values
# isna().sum() checks columns for null values

df = df.fillna(0)
df.isna().sum()
```

```python
# sort by category and record id

df.sort_values(by=["Name", "id"], inplace=True)
df
```

```python
# calculate 3-month rolling average, with at least one observation to produce a non-null value

df["3m_avg"] = df["Value"].rolling(window=3, min_periods=1).mean()
df
```

```python
# format float to display at 0.00 instead of scientific notation

pd.set_option('display.float_format', lambda x: '%.2f' % x)
df
```

```python
# calculate 6-month rolling average, with at least one observation to produce a non-null value

df["6m_avg"] = df["Value"].rolling(window=6, min_periods=1).mean()
df
```

```python
# reset dataframe index, print truncated or scrollable element

df = df.reset_index(drop=True)
print(df.to_string())
```
