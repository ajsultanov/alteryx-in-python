# Challenge #11: Identify Logical Groups

>The use case:

>A manufacturing company receives customer complaint data on a daily basis from their call centers about the medical parts they distribute to their customers. The company monitors these comments to understand which parts and part groups have the highest complaint rate. This helps the company prioritize which parts to focus on from a development standpoint.

>In this exercise, take the customer complaint data and **identify which bucket the complaint falls within. The complaint can fall into multiple buckets and needs to be flagged as these complaints take highest priority. Create an aggregate view of which buckets or bucket pairings have the highest # of complaints.**

>This is only a subset of data so all records will not be assigned to buckets and can be ignored.

```python
import pandas as pd
import numpy as np

# import and preview output

output = pd.read_csv('output.csv', index_col = False)
df_output = pd.DataFrame(output)
df_output.head()
```

```python
# import and preview input 1

input1 = pd.read_csv('input1.csv', index_col = False)
df1 = pd.DataFrame(input1)
max_str = df1.Field_6.str.len().max()
pd.options.display.max_colwidth = max_str
df1
```

```python
# import and preview input 2

input2 = pd.read_csv('input2.csv', index_col = False)
df2 = pd.DataFrame(input2)
df2.head()
```

```python
df1['Field_6'] = df1['Field_6'].str.lower()
df1
```

```python
df1['beep'] = df1['Field_6'].str.match('.*beep.*')
df1['screen'] = df1['Field_6'].str.match('.*screen.*')
df1['trigger'] = df1['Field_6'].str.match('.*trigger.*')
df1
```

```python
df1['beep'].replace(to_replace=True, value='Tones', inplace=True)
df1['beep'].replace(to_replace=False, value=np.nan, inplace=True)
df1['screen'].replace(to_replace=True, value='Screen', inplace=True)
df1['screen'].replace(to_replace=False, value=np.nan, inplace=True)
df1['trigger'].replace(to_replace=True, value='Trigger', inplace=True)
df1['trigger'].replace(to_replace=False, value=np.nan, inplace=True)
df1
```

```python
df1['Bucket(s)'] = df1[['beep', 'screen', 'trigger']].apply(lambda x: ','.join(x.dropna().astype(str).values), axis=1)
df1
```

```python
df1 = df1[df1['Bucket(s)'] != '']
df1
```

```python
df = df1.groupby(by='Bucket(s)').count()
df = df.drop(['beep', 'screen', 'trigger'], axis=1)
df = df.rename({'Field_6': 'Count'}, axis=1)
df = df.reset_index()
df
```
